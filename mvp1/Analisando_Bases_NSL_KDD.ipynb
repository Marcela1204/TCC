```python
columns = [
'duration','protocol_type','service','flag','src_bytes','dst_bytes',
'land','wrong_fragment','urgent','hot','num_failed_logins',
'logged_in','num_compromised','root_shell','su_attempted','num_root',
'num_file_creations','num_shells','num_access_files','num_outbound_cmds',
'is_host_login','is_guest_login','count','srv_count','serror_rate',
'srv_serror_rate','rerror_rate','srv_rerror_rate','same_srv_rate',
'diff_srv_rate','srv_diff_host_rate','dst_host_count','dst_host_srv_count',
'dst_host_same_srv_rate','dst_host_diff_srv_rate',
'dst_host_same_src_port_rate','dst_host_srv_diff_host_rate',
'dst_host_serror_rate','dst_host_srv_serror_rate',
'dst_host_rerror_rate','dst_host_srv_rerror_rate','label','difficulty'
]

df.columns = columns
```

```python
# Coluna Inútil
df = df.drop(columns=['difficulty'])
df = df.drop(columns=['hot'])
df = df.drop(columns=['logged_in'])
```

```python
print(df['label'].value_counts())
```

```python
df['label'].value_counts().plot(kind='bar')
plt.title("Distribuição de ataques")
plt.show()
```

```python
def map_attack(label):
    if label == 'normal':
        return 'normal'
    elif label in ['neptune','back','land','pod','smurf','teardrop']:
        return 'dos'
    elif label in ['ipsweep','nmap','portsweep','satan']:
        return 'probe'
    elif label in ['ftp_write','guess_passwd','imap','multihop','phf','spy','warezclient','warezmaster']:
        return 'r2l'
    else:
        return 'u2r'

df['label'] = df['label'].apply(map_attack)
```

```python
from sklearn.preprocessing import LabelEncoder

categorical_cols = ['protocol_type', 'service', 'flag']

le = LabelEncoder()
for col in categorical_cols:
    df[col] = le.fit_transform(df[col])
```

```python
''' APRENDIZADO SUPERVISIONADO => “Aprenda a prever y usando X”
X (features) = informações da conexão; tudo que o modelo pode usar pra decidir
y (label) = tipo de tráfego (ataque ou normal); a resposta correta
Separamos X e y porque o modelo precisa aprender a prever o resultado (y) com base nas informações (X).'''

X = df.drop('label', axis=1)
y = df['label']
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(X.corr(), cmap='coolwarm')
plt.title("Matriz de Correlação")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.scatter(X_pca[:,0], X_pca[:,1], c=pd.factorize(y)[0], cmap='viridis')
plt.title("PCA - Visualização")
plt.show()
```

```python
import numpy as np

labels = np.unique(y)
colors = plt.cm.viridis(np.linspace(0, 1, len(labels)))

for i, label in enumerate(labels):
    plt.scatter(
        X_pca[y == label, 0],
        X_pca[y == label, 1],
        label=label,
        color=colors[i]
    )

plt.xlabel("Componente Principal 1")
plt.ylabel("Componente Principal 2")
plt.title("PCA - Visualização das Classes")
plt.legend()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3)
```

```python
rf = RandomForestClassifier()
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)

print("Random Forest")
print(classification_report(y_test, y_pred))
```

```python
# Correlação com a variável 'attack'
corr_with_attack = df_numeric.corr()['attack'].sort_values(ascending=False)

print("Top 10 features mais correlacionados com ataques:")
print(corr_with_attack.head(11))  # incluindo ela mesma

# Matriz de correlação geral (opcional, pode ser grande)
plt.figure(figsize=(14, 10))
sns.heatmap(df_numeric.corr(), cmap='coolwarm', center=0, square=True,
            linewidths=0.1, cbar_kws={"shrink": 0.8})
plt.title('Matriz de Correlação - NSL-KDD')
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import time

# Usar dados padronizados (sem PCA primeiro, para identificar features importantes)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

# Dicionário de modelos
modelos = {
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'KNN': KNeighborsClassifier(n_neighbors=5),
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'SVM (linear)': SVC(kernel='linear', random_state=42)
}

resultados = []

print("="*95)
print(f"{'Modelo':<20} {'Acurácia':<10} {'Precisão':<10} {'Recall':<10} {'F1-Score':<10} {'Tempo treino(s)':<10} {'Tempo teste(s)':<10}")
print("="*95)

for nome, modelo in modelos.items():
    start_train = time.time()
    modelo.fit(X_train, y_train)
    end_train = time.time() # Corrected: Call time.time() to get the timestamp
    start_test = time.time()
    y_pred = modelo.predict(X_test)
    end_test = time.time() # Corrected: Call time.time() to get the timestamp

    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred)
    rec = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    resultados.append({
        'Modelo': nome,
        'Acurácia': acc,
        'Precisão': prec,
        'Recall': rec,
        'F1-Score': f1,
        'Tempo treino(s)': end_train - start_train,
        'tempo teste(s)': end_test - start_test
    })

    print(f"{nome:<20} {acc:<10.4f} {prec:<10.4f} {rec:<10.4f} {f1:<10.4f} {end_train-start_train:<15.4f} {end_test-start_test:<15.4f}")

# Dataframe com resultados
df_resultados = pd.DataFrame(resultados)
print("\nMelhor modelo por F1-Score:")
print(df_resultados.loc[df_resultados['F1-Score'].idxmax()])
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
import os
import time
import warnings
warnings.filterwarnings('ignore')

# Download do dataset NSL-KDD
print("Baixando dataset NSL-KDD...")
path = kagglehub.dataset_download("hassan06/nslkdd")
print(f"Dataset baixado em: {path}")

# Listar arquivos disponíveis
print("\nArquivos disponíveis:")
for file in os.listdir(path):
    print(f"  - {file}")

# Colunas do NSL-KDD (41 features + label + difficulty)
colunas = [
    'duration', 'protocol_type', 'service', 'flag', 'src_bytes', 'dst_bytes',
    'land', 'wrong_fragment', 'urgent', 'hot', 'num_failed_logins',
    'logged_in', 'num_compromised', 'root_shell', 'su_attempted', 'num_root',
    'num_file_creations', 'num_shells', 'num_access_files', 'num_outbound_cmds',
    'is_host_login', 'is_guest_login', 'count', 'srv_count', 'serror_rate',
    'srv_serror_rate', 'rerror_rate', 'srv_rerror_rate', 'same_srv_rate',
    'diff_srv_rate', 'srv_diff_host_rate', 'dst_host_count', 'dst_host_srv_count',
    'dst_host_same_srv_rate', 'dst_host_diff_srv_rate', 'dst_host_same_src_port_rate',
    'dst_host_srv_diff_host_rate', 'dst_host_serror_rate', 'dst_host_srv_serror_rate',
    'dst_host_rerror_rate', 'dst_host_srv_rerror_rate', 'label', 'difficulty'
]

# Carregar KDDTrain+ (treinamento)
print("\nCarregando KDDTrain+.txt...")
train_file = os.path.join(path, 'KDDTrain+.txt')
df_train = pd.read_csv(train_file, names=colunas, sep=',') # Fixed separator to comma

# Carregar KDDTest+ (teste - opcional, para validação final)
print("Carregando KDDTest+.txt...")
test_file = os.path.join(path, 'KDDTest+.txt')
df_test = pd.read_csv(test_file, names=colunas, sep=',') # Fixed separator to comma

# Dropping columns as requested by the user
df_train = df_train.drop(columns=['difficulty', 'hot', 'logged_in'])
df_test = df_test.drop(columns=['difficulty', 'hot', 'logged_in'])

print(f"\nShape treino: {df_train.shape}")
print(f"Shape teste: {df_test.shape}")

# Clean the 'label' column to remove trailing dots and whitespace
df_train['label'] = df_train['label'].astype(str).str.replace('.', '', regex=False).str.strip()
df_test['label'] = df_test['label'].astype(str).str.replace('.', '', regex=False).str.strip()

print("\nDistribuição das labels originais (após limpeza):")
print(df_train['label'].value_counts())

# Criar target binário: normal (0) ou ataque (1)
df_train['attack'] = df_train['label'].apply(lambda x: 0 if x == 'normal' else 1)
df_test['attack'] = df_test['label'].apply(lambda x: 0 if x == 'normal' else 1)

# Ver distribuição das classes
print("\nDistribuição das classes (treino):")
print(df_train['attack'].value_counts())
print(f"Normal: {df_train['attack'].value_counts().get(0, 0)} ({df_train['attack'].value_counts().get(0, 0)/len(df_train)*100:.2f}%)") # Using .get(0,0) for robustness
print(f"Ataque: {df_train['attack'].value_counts().get(1, 0)} ({df_train['attack'].value_counts().get(1, 0)/len(df_train)*100:.2f}%)") # Using .get(1,0) for robustness
```

```python
# Selecionar apenas colunas numéricas
df_train_numeric = df_train.select_dtypes(include=[np.number])
df_train_numeric = df_train_numeric.drop('difficulty', axis=1, errors='ignore')

# Correlação com a variável 'attack'
corr_with_attack = df_train_numeric.corr()['attack'].sort_values(ascending=False)

print("\n" + "="*60)
print("TOP 10 FEATURES MAIS CORRELACIONADAS COM ATAQUES:")
print("="*60)
for i, (feature, corr) in enumerate(corr_with_attack.head(11).items(), 1):
    if feature != 'attack':
        print(f"{i:2d}. {feature:35s} {corr:+.4f}")

print("\n" + "="*60)
print("TOP 10 FEATURES MENOS CORRELACIONADAS (ou anti-correlacionadas):")
print("="*60)
for i, (feature, corr) in enumerate(corr_with_attack.tail(10).items(), 1):
    print(f"{i:2d}. {feature:35s} {corr:+.4f}")

# Heatmap das top 20 features correlacionadas com o ataque
top_features = corr_with_attack.head(21).index.tolist()  # incluindo 'attack'
plt.figure(figsize=(12, 10))
sns.heatmap(df_train_numeric[top_features].corr(),
            cmap='coolwarm', center=0,
            annot=True, fmt='.2f',
            square=True,
            linewidths=0.5,
            cbar_kws={"shrink": 0.8})
plt.title('Matriz de Correlação - Top 20 Features (incluindo target)')
plt.tight_layout()
plt.show()
```

```python
# Preparar dados para PCA (apenas features numéricas, sem target)
X = df_train_numeric.drop('attack', axis=1)
y = df_train['attack'].values

# Padronizar (PCA é sensível a escala)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Aplicar PCA
print("\n" + "="*60)
print("ANÁLISE DE PCA:")
print("="*60)

# PCA com 95% da variância
pca_95 = PCA(n_components=0.95)
X_pca_95 = pca_95.fit_transform(X_scaled)

print(f"Dimensão original: {X_scaled.shape[1]} features")
print(f"Dimensão após PCA (95% variância): {X_pca_95.shape[1]} componentes")
print(f"Variância explicada acumulada: {np.cumsum(pca_95.explained_variance_ratio_)[-1]:.2%}")

# PCA com componentes fixos (para visualização 2D)
pca_2d = PCA(n_components=2)
X_pca_2d = pca_2d.fit_transform(X_scaled)

# Plot 2D PCA colorido por classe
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X_pca_2d[y==0, 0], X_pca_2d[y==0, 1], c='blue', alpha=0.5, s=1, label='Normal')
plt.scatter(X_pca_2d[y==1, 0], X_pca_2d[y==1, 1], c='red', alpha=0.5, s=1, label='Ataque')
plt.xlabel('Primeira Componente Principal')
plt.ylabel('Segunda Componente Principal')
plt.title('PCA 2D - NSL-KDD')
plt.legend()
plt.grid(True, alpha=0.3)

# Variância explicada acumulada
plt.subplot(1, 2, 2)
cumulative_variance = np.cumsum(pca_95.explained_variance_ratio_)
plt.plot(range(1, len(cumulative_variance)+1), cumulative_variance, 'bo-', linewidth=2)
plt.axhline(y=0.95, color='r', linestyle='--', label='95% variância')
plt.xlabel('Número de Componentes')
plt.ylabel('Variância Explicada Acumulada')
plt.title('PCA - Variância Explicada Acumulada')
plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()

print(f"\nPara manter 95% da variância, você precisa de {pca_95.n_components_} componentes")
```

```python
# Usar dados padronizados (X_scaled já criado)
X_train, X_val, y_train, y_val = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

# Dicionário de modelos
modelos = {
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'KNN (k=5)': KNeighborsClassifier(n_neighbors=5),
    'KNN (k=3)': KNeighborsClassifier(n_neighbors=3),
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'SVM (linear)': SVC(kernel='linear', random_state=42)
}

resultados = []

print("\n" + "="*80)
print(f"{'Modelo':<20} {'Acurácia':<10} {'Precisão':<10} {'Recall':<10} {'F1-Score':<10} {'Tempo treino(s)':<10}")
print("="*80)

for nome, modelo in modelos.items():
    start = time.time()
    modelo.fit(X_train, y_train)
    y_pred = modelo.predict(X_val)
    end = time.time()

    acc = accuracy_score(y_val, y_pred)
    prec = precision_score(y_val, y_pred)
    rec = recall_score(y_val, y_pred)
    f1 = f1_score(y_val, y_pred)

    resultados.append({
        'Modelo': nome,
        'Acurácia': acc,
        'Precisão': prec,
        'Recall': rec,
        'F1-Score': f1,
        'Tempo treino(s)': end - start
    })

    print(f"{nome:<20} {acc:<10.4f} {prec:<10.4f} {rec:<10.4f} {f1:<10.4f} {end-start:<15.4f}")

# Dataframe com resultados
df_resultados = pd.DataFrame(resultados)
df_resultados = df_resultados.sort_values('F1-Score', ascending=False)

print("\n" + "="*60)
print("RANKING DOS MODELOS (por F1-Score):")
print("="*60)
print(df_resultados.to_string(index=False))

# Gráfico comparativo
plt.figure(figsize=(12, 6))
df_resultados_melted = df_resultados.melt(id_vars=['Modelo'],
                                          value_vars=['Acurácia', 'Precisão', 'Recall', 'F1-Score'],
                                          var_name='Métrica',
                                          value_name='Valor')
sns.barplot(data=df_resultados_melted, x='Modelo', y='Valor', hue='Métrica')
plt.title('Comparação de Desempenho dos Modelos')
plt.xticks(rotation=45, ha='right')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score # Alterado de f1_score para accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns

# Certifique-se de que X_scaled, y e 'importances' (do Random Forest) estejam disponíveis
# 'importances' foi calculado no Teste 3, célula N831OsMCHmx8

if 'importances' not in globals():
    print("Variável 'importances' não encontrada. Execute as células anteriores.")
    # Recriar importances para garantir que o código seja executável se for rodado isoladamente
    # (Este é um fallback, o ideal é que o notebook seja executado em ordem)
    try:
        rf_temp = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
        rf_temp.fit(X_scaled, y)
        feature_names = X.columns
        importances = pd.DataFrame({
            'feature': feature_names,
            'importance': rf_temp.feature_importances_
        }).sort_values('importance', ascending=False)
    except Exception as e:
        print(f"Erro ao recriar importances: {e}. Verifique se X_scaled e y estão definidos.")
        exit()

# Armazenar os resultados
results = []

# Loop para testar de 5 a 38 features
for n_features in range(5, 39):
    # Selecionar as top n_features com base na importância
    top_features_names = importances['feature'].head(n_features).tolist()
    # Obter os índices das colunas no DataFrame original (X) para X_scaled
    feature_indices = [X.columns.get_loc(col) for col in top_features_names]

    X_selected = X_scaled[:, feature_indices]

    # Dividir os dados em treino e teste
    X_train_nf, X_test_nf, y_train_nf, y_test_nf = train_test_split(
        X_selected, y, test_size=0.2, random_state=42, stratify=y
    )

    # Treinar o modelo Decision Tree
    dt_model = DecisionTreeClassifier(random_state=42)
    dt_model.fit(X_train_nf, y_train_nf)

    # Fazer previsões e calcular a acurácia
    y_pred_nf = dt_model.predict(X_test_nf)
    accuracy = accuracy_score(y_test_nf, y_pred_nf) # Alterado de f1 para accuracy

    results.append({'n_features': n_features, 'accuracy_score': accuracy}) # Alterado de f1_score para accuracy_score

# Converter resultados para DataFrame para visualização
df_results = pd.DataFrame(results)

print("\n" + "="*60)
print("DESEMPENHO DA DECISION TREE COM DIFERENTES NÚMEROS DE FEATURES")
print("="*60)
print(df_results.to_string(index=False))

# Plotar os resultados
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_results, x='n_features', y='accuracy_score', marker='o') # Alterado de f1_score para accuracy_score
plt.title('Acurácia da Decision Tree vs. Número de Features') # Alterado de F1-Score para Acurácia
plt.xlabel('Número de Features')
plt.ylabel('Acurácia') # Alterado de F1-Score para Acurácia
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(range(5, 39, 1)) # Mostrar todos os números de features no eixo X
plt.tight_layout()
plt.show()

# Encontrar o número de features com a maior acurácia
best_performance = df_results.loc[df_results['accuracy_score'].idxmax()] # Alterado de f1_score para accuracy_score
print(f"\nMelhor Acurácia: {best_performance['accuracy_score']:.4f} com {int(best_performance['n_features'])} features.") # Alterado de F1-Score para Acurácia
```
